```python
final=cordinate.merge(temp,left_on='District',right_on='District name').drop(columns=['District name'])
```

```python
final.head()
```

```python
final['sex_ratio']=round((final['Female']/final['Male'])*100)
```

```python
final['literacy_rate']=round((final['Literate']/final['Population'])*100)
```

```python
final['Graduate_rate']=round(final['Graduate_Education']/final['Literate']*100)
```

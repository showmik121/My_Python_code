কেন arange এর চেয়ে linspace ভালো?
আপনি বলতে পারেন, "আমি তো np.arange দিয়েও সংখ্যা তৈরি করতে পারি।" কিন্তু তাদের মধ্যে একটি বড় পার্থক্য আছে:

np.arange: এখানে আপনি বলেন সংখ্যাগুলোর মধ্যে ব্যবধান (Step) কত হবে। (যেমন: -১০ থেকে ১০ এর মধ্যে ০.৫ করে ব্যবধান)। এখানে আপনি আগে থেকে জানেন না মোট কয়টি সংখ্যা তৈরি হবে।

np.linspace: এখানে আপনি বলেন আপনার মোট কয়টি সংখ্যা (Count) লাগবে। (যেমন: -১০ থেকে ১০ এর মধ্যে ঠিক ১০০টি সংখ্যা)। গ্রাফ আঁকার সময় আমাদের সাধারণত পয়েন্টের সংখ্যা নির্দিষ্ট রাখা বেশি সুবিধাজনক হয়।

৩. শেষ সীমানা (Endpoints) নিশ্চিত করা
np.arange ব্যবহার করলে অনেক সময় শেষ সংখ্যাটি (যেমন ১০) বাদ পড়ে যেতে পারে। কিন্তু np.linspace গ্যারান্টি দেয় যে আপনার দেওয়া শুরু (-১০) এবং শেষ (১০) উভয় সংখ্যাই লিস্টে থাকবে।

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
x
```

np.meshgrid মূলত আপনার ১ডি (1D) অ্যারেগুলোকে ২ডি (2D) গ্রিডে রূপান্তর করে। সহজ ভাষায়, এটি একটি গ্রাফ পেপারের মতো কাজ করে যেখানে প্রতিটি $x$ এবং $y$ এর সংমিশ্রণে একটি স্থানাঙ্ক বা পয়েন্ট তৈরি হয়।

```python
xx,yy=np.meshgrid(x,y)
```

```python
yy.shape
```

```python
z=xx**2 + yy**2
```

```python
z.shape
```

```python
fig=plt.figure(figsize=(12,8))
```

```python
ax=plt.subplot(projection='3d')
ax.plot_surface(xx,yy,z)
```

```python
ax=plt.subplot(projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')
```

```python
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

3D -> 2D

Contour plot

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

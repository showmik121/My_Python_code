```python
np.arange(season.shape[0])
```

```python
x = np.arange(season.shape[0])
w = 0.25 # বারের চওড়া

# ২. বারগুলো প্লট করা (পজিশন শিফট করে)
plt.bar(x - w, season['2015'], width=w, label='2015') # বামে
plt.bar(x,     season['2016'], width=w, label='2016') # মাঝে
plt.bar(x + w, season['2017'], width=w, label='2017') # ডানে
```

সহজ কথায়, plt.xticks() এর কাজ হলো আপনার গ্রাফের নিচের অক্ষের (X-axis) নম্বরগুলোকে নাম দিয়ে বদলে দেওয়া।

```python
x = np.arange(season.shape[0])
w = 0.25 # বারের চওড়া

# ২. বারগুলো প্লট করা (পজিশন শিফট করে)
plt.bar(x - w, season['2015'], width=w, label='2015') # বামে
plt.bar(x,     season['2016'], width=w, label='2016') # মাঝে
plt.bar(x + w, season['2017'], width=w, label='2017') # ডানে

plt.xticks(x,season['batsman'])
plt.grid()
plt.show()
```

stacked bar chat

```python
import matplotlib.pyplot as plt

# প্রথম বার (সবার নিচে থাকবে ২০১৭)
plt.bar(season['batsman'], season['2017'], label='2017')

# দ্বিতীয় বার (২০১৭ এর ওপর বসবে ২০১৬)
plt.bar(season['batsman'], season['2016'], bottom=season['2017'], label='2016')

# তৃতীয় বার (২০১৬ এবং ২০১৭ এর যোগফলের ওপর বসবে ২০১৫)
plt.bar(season['batsman'], season['2015'], bottom=season['2016'] + season['2017'], label='2015')

plt.xticks(rotation='vertical')
plt.legend()
plt.show()
```

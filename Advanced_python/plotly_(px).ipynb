plotly-"Next Level" ভিজ্যুয়ালাইজেশন লাইব্রেরি


plotly দিয়ে আমরা এমন গ্রাফ তৈরি করি যা Interactive—অর্থাৎ মাউস নাড়ালে ডেটা দেখা যায়, জুম করা যায় এবং ফিল্টার করা যায়।

Plotly Express (px): এটি নতুনদের জন্য। মাত্র এক লাইনে খুব সুন্দর গ্রাফ তৈরি করা যায়। (Seaborn-এর মতো সহজ)।

Plotly Graph Objects (go): এটি অ্যাডভান্স ইউজারদের জন্য। গ্রাফের প্রতিটি খুঁটিনাটি (কাস্টমাইজেশন) নিয়ন্ত্রণ করতে এটি ব্যবহার হয়।

Dash (The Power of Plotly)
আপনি যখন Plotly শিখবেন, তখন Dash সম্পর্কেও জানবেন। Dash ব্যবহার করে আপনি পাইথন দিয়েই পুরো একটি Web Dashboard বা অ্যাপ তৈরি করে ফেলতে পারেন, যেখানে কোনো HTML/CSS জানার প্রয়োজন নেই।

```python
trace1=go.Scatter(x=temp['lifeExp'],y=temp['gdpPercap'])
data=[trace1] #list
layout=go.Layout(title='life exp vs gdp per capital for 2007',xaxis={'title':'life exp'},yaxis={'title':'gpd'})
fig=go.Figure(data,layout)
fig.show(renderer='browser')
```

```python
trace1=go.Scatter(x=temp['lifeExp'],y=temp['gdpPercap'],mode='markers')
data=[trace1] #list
layout=go.Layout(title='life exp vs gdp per capital for 2007',xaxis={'title':'life exp'},yaxis={'title':'gpd'})
fig=go.Figure(data,layout)
fig.show(renderer='browser')
```

```python
trace1=go.Scatter(x=temp['lifeExp'],y=temp['gdpPercap'])
trace2=go.Scatter(x=[0,1,2],y=[0,90,30000],mode='lines')
data=[trace1,trace2] #list
layout=go.Layout(title='life exp vs gdp per capital for 2007',xaxis={'title':'life exp'},yaxis={'title':'gpd'})
fig=go.Figure(data,layout)
fig.show(renderer='browser')
```
